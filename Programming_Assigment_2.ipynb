# A. REPRODUCIBLE NORMALIZATION PROBLEM

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5,5))
Mean = np.mean(X)
STD = np.std(X)
X_normalized = (X-Mean)/STD
NormM = np.mean(X_normalized)
NormS = np.std(X_normalized)
print("X-Normalized:")
print(X_normalized)
print(" ")
print("X:")
print(X)
print(" ")
print("Mean:")
print(Mean)
print(" ")
print("Standard Deviation:")
print(STD)
print(" ")
print("X-Normalized Mean:")
print(NormM)
print(" ")
print("X-Normalized STD:")
print(NormS)
np.save("X_normalized.npy", X_normalized)
```

# B. CUBES DIVISIBLE BY 4 PROBLEM

```python
A = np.arange(1,101)
B = A*A*A
C = B.reshape(10,10)
div_by_4 = C[C % 4 ==0]
print("Shape of C")
print(C.shape)
print(" ")
print("div_by_4:")
print(div_by_4)
print(" ")
print("Number of Elemets:")
print(len(div_by_4))
np.save("div_by_4.npy", div_by_4)
```

# C. ABOVE-MEAN SQUARES

```python
S = np.arange(1, 37)
S = S*S
S = S.reshape(6,6)
S_mean = S.mean()
above_mean = S[S > S_mean]
print ("S:")
print (S)
print (" ")
print ("Mean:")
print (S_mean)
print (" ")
print ("Above Mean:")
print (above_mean)
print(" ")
print ("Number of Elements:")
print (len(above_mean))
np.save("above_mean.npy", above_mean)
```
